# Ex 16.1:

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import time
```

#### Load dataset

```python
mnist = keras.datasets.mnist
fashion = keras.datasets.fashion_mnist

(x_train_m, y_train_m), (x_test_m, y_test_m) = mnist.load_data()
(x_train_f, y_train_f), (x_test_f, y_test_f) = fashion.load_data()
```

#### Normalize

```python
x_train_m = x_train_m / 255.0
x_test_m = x_test_m / 255.0
x_train_f = x_train_f / 255.0
x_test_f = x_test_f / 255.0
```

#### # reshape for CNN

```python
x_train_m = x_train_m.reshape(-1,28,28,1)
x_test_m = x_test_m.reshape(-1,28,28,1)

x_train_f = x_train_f.reshape(-1,28,28,1)
x_test_f = x_test_f.reshape(-1,28,28,1)
```

```python
def build_model():
    model = keras.Sequential([
        layers.Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)),
        layers.MaxPooling2D((2,2)),
        layers.Conv2D(64,(3,3),activation='relu'),
        layers.MaxPooling2D((2,2)),
        layers.Flatten(),
        layers.Dense(128,activation='relu'),
        layers.Dense(10,activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model
```

#### Train MNIST

```python
model_m = build_model()

start = time.time()
model_m.fit(x_train_m, y_train_m, epochs=5, validation_split=0.1)
mnist_time = time.time() - start

mnist_loss, mnist_acc = model_m.evaluate(x_test_m, y_test_m)

print("MNIST accuracy:", mnist_acc)
print("MNIST training time:", mnist_time)
```

#### Train Fashion-MNIST

```python
model_f = build_model()

start = time.time()
model_f.fit(x_train_f, y_train_f, epochs=5, validation_split=0.1)
fashion_time = time.time() - start

fashion_loss, fashion_acc = model_f.evaluate(x_test_f, y_test_f)

print("Fashion-MNIST accuracy:", fashion_acc)
print("Fashion-MNIST training time:", fashion_time)
```



*  The model achieves a higher accuracy on MNIST (99%) compared to Fashion-MNIST (90%)


*  It is more challenging to classify Fashion-MNIST items correctly, as clothes have alike shapes.

```python
def build_model_with_4096():
    model = keras.Sequential([
        layers.Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)),
        layers.MaxPooling2D((2,2)),
        layers.Conv2D(64,(3,3),activation='relu'),
        layers.MaxPooling2D((2,2)),
        layers.Flatten(),

        layers.Dense(4096,activation='relu'),   # NEW LAYER
        layers.Dense(128,activation='relu'),

        layers.Dense(10,activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model


model_big = build_model_with_4096()

start = time.time()
model_big.fit(x_train_f, y_train_f, epochs=5, validation_split=0.1)
big_time = time.time() - start

loss, acc = model_big.evaluate(x_test_f, y_test_f)

print("Accuracy with 4096 Dense:", acc)
print("Training time:", big_time)
```
